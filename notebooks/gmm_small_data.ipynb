```python
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
X, true_labels = make_blobs(
    n_samples=500,
    centers=3,
    cluster_std=[1.0, 1.5, 0.8],
    random_state=42,
)

plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], s=25)
plt.title("Sample data (true labels hidden)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
gmm = GaussianMixture(
    n_components=3,
    covariance_type="full",
    random_state=42
)

gmm.fit(X)
```

```python
cluster_labels = gmm.predict(X)
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=cluster_labels, cmap="viridis", s=25, alpha=0.6)
plt.title("Clusters found by GMM")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
membership_probs = gmm.predict_proba(X)

print("Probability of the first 5 points: ")
print(membership_probs[:5].round(3))

print("\nLearned component weights: ")
print(gmm.weights_.round(3))
```

```python
print(f"Shape of all the points: {membership_probs.shape}")
```

```python
unique_vals, counts = np.unique(cluster_labels, return_counts=True)

for unique_val, count in zip(unique_vals, counts):
    print(f"Clusters {unique_val}: {count} points")
```

```python
component_counts = range(1, 7)
aic_scores = []
bic_scores = []

for n in component_counts:
    candidate_gmm = GaussianMixture(
        n_components=n,
        covariance_type="full",
        random_state=42
    )

    candidate_gmm.fit(X)

    aic_scores.append(candidate_gmm.aic(X))
    bic_scores.append(candidate_gmm.bic(X))
    
plt.figure(figsize=(14, 7))
plt.plot(component_counts, aic_scores, marker="o", label="AIC")
plt.plot(component_counts, bic_scores, marker="o", label="BIC")
plt.xlabel("Number of components")
plt.ylabel("Score (lower) is better")
plt.legend()
plt.grid(alpha=0.7)
plt.show()
```

```python
covariance_types = ["full", "tied", "diag", "spherical"]
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

for ax, cov_type in zip(axes.flatten(), covariance_types):
    model = GaussianMixture(
        n_components=3,
        covariance_type=cov_type,
        random_state=42
    )

    model.fit(X)
    label= model.predict(X)
    bic = model.bic(X)

    ax.scatter(X[:,0], X[:,1], c=label, cmap="viridis", s=20)
    ax.set_title(f"{cov_type} covariance — BIC: {bic:.1f}")
    ax.set_xlabel("Feature 1")
    ax.set_ylabel("Feature 2")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

for ax, cov_type in zip(axes.flatten(), covariance_types):
    model = GaussianMixture(
        n_components=3,
        covariance_type=cov_type,
        random_state=42
    )

    model.fit(X)
    label= model.predict(X)
    aic = model.aic(X)

    ax.scatter(X[:,0], X[:,1], c=label, cmap="viridis", s=20)
    ax.set_title(f"{cov_type} covariance — AIC: {aic:.1f}")
    ax.set_xlabel("Feature 1")
    ax.set_ylabel("Feature 2")

plt.tight_layout()
plt.show()
```

```python
best_gmm = GaussianMixture(
    n_components=3,
    covariance_type="spherical",
    random_state=42
)

best_gmm.fit(X)

best_label = best_gmm.predict(X)
best_aic = best_gmm.aic(X)
best_bic = best_gmm.bic(X)
best_membership_probs = best_gmm.predict_proba(X)

print("Probability of first 5 components: ")
print(best_membership_probs[:5].round(3))

print("\nLearned component weights: ")
print(best_gmm.weights_.round(3))

print("\nComponent centers:")
print(best_gmm.means_.round(2))

print("\nComponent variances:")
print(best_gmm.covariances_.round(3))
```

```python
plt.figure(figsize=(14, 7))
plt.scatter(X[:,0], X[:,1], c=best_label, cmap="viridis", s=25)
plt.title("Clusters found by GMM")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.tight_layout()
plt.show()
```

```python
results = []

for n in range(1, 7):
    for cov_type in ["full", "tied", "diag", "spherical"]:
        candidate = GaussianMixture(
            n_components=n,
            covariance_type=cov_type,
            random_state=42
        )

        candidate.fit(X)
        results.append({
            "n_components": n,
            "covariance_type":cov_type,
            "AIC": candidate.aic(X),
            "BIC":candidate.bic(X)
        })

best_aic = min(results, key=lambda row: row["AIC"])
best_bic = min(results, key=lambda row: row["BIC"])

print("Best by aic: ", best_aic)
print("Best by bic: ", best_bic)
    
```

```python
uniques, counts = np.unique(best_label, return_counts=True)

for unique, count in zip(uniques, counts):
    print(f"Cluster {unique}: {count} points.")
```
