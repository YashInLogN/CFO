```python
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
X, true_labels = make_blobs(
    n_samples=500,
    centers=3,
    cluster_std=[1.0, 1.5, 0.8],
    random_state=42,
)

plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], s=25)
plt.title("Sample data (true labels hidden)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
gmm = GaussianMixture(
    n_components=3,
    covariance_type="full",
    random_state=42
)

gmm.fit(X)
```

```python
cluster_labels = gmm.predict(X)
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=cluster_labels, cmap="viridis", s=25, alpha=0.6)
plt.title("Clusters found by GMM")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
membership_probs = gmm.predict_proba(X)

print("Probability of the first 5 points: ")
print(membership_probs[:5].round(3))

print("\nLearned component weights: ")
print(gmm.weights_.round(3))
```

```python
print(f"Shape of all the points: {membership_probs.shape}")
```

```python
unique_vals, counts = np.unique(cluster_labels, return_counts=True)

for unique_val, count in zip(unique_vals, counts):
    print(f"Clusters {unique_val}: {count} points")
```

```python
component_counts = range(1, 7)
aic_scores = []
bic_scores = []

for n in component_counts:
    candidate_gmm = GaussianMixture(
        n_components=n,
        covariance_type="full",
        random_state=42
    )

    candidate_gmm.fit(X)

    aic_scores.append(candidate_gmm.aic(X))
    bic_scores.append(candidate_gmm.bic(X))
    
plt.figure(figsize=(14, 7))
plt.plot(component_counts, aic_scores, marker="o", label="AIC")
plt.plot(component_counts, bic_scores, marker="o", label="BIC")
plt.xlabel("Number of components")
plt.ylabel("Score (lower) is better")
plt.legend()
plt.grid(alpha=0.7)
plt.show()
```
